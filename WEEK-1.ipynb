```python
x = np.array([21, 64, 86, 22, 74, 55, 81, 79, 90, 89])
y = np.array([21, 7, 3, 45, 10, 29, 55, 4, 37, 18])
greater = np.where(x > y)
equal = np.where(x == y)
print(greater)
equal
```

2. Consider the following matrices : <br>
A= ((1, 2, 3), (4, 5, 6), (7, 8, 10)) and B = ((7, 8, 10) ,(4, 5, 6), (1, 2, 3))<br>
Write a python program to perform the following using Numeric Python (numpy).<br>
i) Add and Subtract of the Matrix A and B, print the resultant matrix C for add and E for subtract.
<br>ii) Compute the sum of all elements of Matrix A, sum of each column of Matrix B and sum of each row of Matrix C
<br>iii) Product of two matrices A and B, and print the resultant matrix D
<br>iv) Sort the elements of resultant matrix C and print the resultant Matrix E.
<br>v) Transpose the Matrix E and print the result

```python
A = np.array([[1, 2, 3], [4, 5, 6], [7, 8, 10]])
B = np.array([[7, 8, 10], [4, 5, 6], [1, 2, 3]])

print("Matrix A:\n", A)
print("\nMatrix B:\n", B)
#i
C = A + B
E_sub = A - B
print("Resultant Matrix C (Add):\n", C)
print("Resultant Matrix E (Subtract):\n", E_sub)

# ii
sum_A = np.sum(A)
sum_col_B = np.sum(B, axis=0) 
sum_row_C = np.sum(C, axis=1) 
print("Sum of all elements of Matrix A:", sum_A)
print("Sum of each column of Matrix B:", sum_col_B)
print("Sum of each row of Matrix C:", sum_row_C)

# iii
D = np.dot(A, B) 
print("Resultant Matrix D (Product):\n", D)

# iv
E = np.sort(C)
print("Resultant Matrix E:\n", E)

# v
E_trans = E.T
print("Resultant Transposed Matrix:\n", E_trans)
```
